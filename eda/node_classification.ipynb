```python
def build_graph(edges, nodes):
    nodes = nodes.sort_values('id').reset_index(drop=True)
    id_map = {nid: i for i, nid in enumerate(nodes['id'].values)}

    # features
    x = torch.tensor(
        nodes[['pos_x','pos_y','pos_z']].values,
        dtype=torch.float
    )

    # FIX: aggregate trước
    edge_radius = edges.groupby('node1id')['maxRadiusAvg'].mean().reset_index()

    nodes_merged = nodes.merge(
        edge_radius,
        left_on='id', right_on='node1id', how='left'
    )

    nodes_merged['maxRadiusAvg'] = nodes_merged['maxRadiusAvg'].fillna(0.0)

    # label
    conditions = [
        nodes_merged['maxRadiusAvg'] < 15,
        (nodes_merged['maxRadiusAvg'] >= 15) & (nodes_merged['maxRadiusAvg'] <= 35),
        nodes_merged['maxRadiusAvg'] > 35
    ]

    nodes_merged['vessel_type'] = np.select(conditions, [0,1,2])

    print("Class distribution:", np.bincount(nodes_merged['vessel_type']))

    y = torch.tensor(nodes_merged['vessel_type'].values, dtype=torch.long)

    # edge index
    edge_index = edges[['node1id','node2id']].copy()
    edge_index['node1id'] = edge_index['node1id'].map(id_map)
    edge_index['node2id'] = edge_index['node2id'].map(id_map)
    edge_index = edge_index.dropna()

    edge_index = torch.tensor(edge_index.values.T, dtype=torch.long)

    return Data(x=x, edge_index=edge_index, y=y)
```

```python
train_data = build_graph(BALBc_2_1, BALBc_2_2)
val_data   = build_graph(BALBc_3_1, BALBc_3_2)
test_data  = build_graph(BALBc_1_1, BALBc_1_2)
```

```python
class GCN(nn.Module):
    def __init__(self, in_dim, hidden=128):
        super().__init__()
        self.conv1 = GCNConv(in_dim, hidden)
        self.conv2 = GCNConv(hidden, hidden)
        self.fc = nn.Linear(hidden, 3)

    def forward(self, x, edge_index):
        x = F.relu(self.conv1(x, edge_index))
        x = F.relu(self.conv2(x, edge_index))
        return self.fc(x)
    
train_loader = NeighborLoader(
    train_data,
    num_neighbors=[10,5],
    batch_size=8192,
    shuffle=True
)
```

```python

device = "cuda" if torch.cuda.is_available() else "cpu"
model = GCN(train_data.num_node_features, hidden=128).to(device)
optimizer = torch.optim.AdamW(model.parameters(), lr=3e-3, weight_decay=1e-4)


y_np = train_data.y.cpu().numpy()
counts = np.bincount(y_np, minlength=3)

# class weights (handle imbalance)
weights = 1.0 / (counts + 1e-6)
weights = weights / weights.sum() * 3

class_weights = torch.tensor(weights, dtype=torch.float, device=device)

criterion = nn.CrossEntropyLoss(weight=class_weights)


@torch.no_grad()
def evaluate(model, data):
    model.eval()

    loader = NeighborLoader(
        data,
        num_neighbors=[10, 5],
        batch_size=4096,
        shuffle=False
    )

    all_probs = []
    all_preds = []
    all_y = []

    for batch in loader:
        batch = batch.to(device)

        logits = model(batch.x, batch.edge_index)

        out = logits[:batch.batch_size]
        y = batch.y[:batch.batch_size]

        probs = torch.softmax(out, dim=1)
        preds = probs.argmax(dim=1)

        all_probs.append(probs.cpu())
        all_preds.append(preds.cpu())
        all_y.append(y.cpu())

    probs = torch.cat(all_probs).numpy()
    preds = torch.cat(all_preds).numpy()
    y_true = torch.cat(all_y).numpy()

    acc = accuracy_score(y_true, preds)
    f1 = f1_score(y_true, preds, average='macro')  # quan trọng

    try:
        auc = roc_auc_score(y_true, probs, multi_class='ovr')
    except:
        auc = 0.0

    return acc, f1, auc
```

```python
num_epochs = 10

for epoch in range(num_epochs):
    model.train()
    total_loss = 0.0
    num_batches = 0

    for batch in train_loader:
        batch = batch.to(device)

        optimizer.zero_grad()

        logits = model(batch.x, batch.edge_index)

        out = logits[:batch.batch_size]
        y = batch.y[:batch.batch_size]  
        loss = criterion(out, y)      

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        total_loss += loss.item()
        num_batches += 1

    train_loss = total_loss / max(1, num_batches)

    val_acc, val_f1, val_auc = evaluate(model, val_data)

    print(f"""
Epoch {epoch+1}/{num_epochs}
Loss: {train_loss:.6f}
Val Acc: {val_acc:.4f} | F1: {val_f1:.4f} | AUC: {val_auc:.4f}
""")
```

```python
test_acc, test_f1, test_auc = evaluate(model, test_data)

print("===== TEST RESULT =====")
print(f"Accuracy : {test_acc:.4f}")
print(f"F1-score : {test_f1:.4f}")
print(f"ROC-AUC  : {test_auc:.4f}")
```
