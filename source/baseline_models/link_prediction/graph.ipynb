```python
import re
import matplotlib.pyplot as plt


def plot_training_log(log_file, plot_title="Training Result"):
    epochs = []
    losses = []
    train_acc = []
    valid_acc = []
    test_acc = []

    pattern = re.compile(
        r"Epoch:\s*(\d+),\s*Loss:\s*([\d.]+),\s*Train:\s*([\d.]+)%,\s*Valid:\s*([\d.]+)%,\s*Test:\s*([\d.]+)%"
    )

    with open(log_file, "r", encoding="utf-8", errors="ignore") as f:
        for line in f:
            match = pattern.search(line)

            if match:
                epochs.append(int(match.group(1)))
                losses.append(float(match.group(2)))
                train_acc.append(float(match.group(3)))
                valid_acc.append(float(match.group(4)))
                test_acc.append(float(match.group(5)))

    # Plot
    fig, ax = plt.subplots(1, 2, figsize=(14, 5))

    # Accuracy
    ax[0].plot(epochs, train_acc, marker='o', label='Train')
    ax[0].plot(epochs, valid_acc, marker='s', label='Validation')
    ax[0].plot(epochs, test_acc, marker='^', label='Test')

    ax[0].set_title(f"{plot_title} - Accuracy")
    ax[0].set_xlabel("Epoch")
    ax[0].set_ylabel("Accuracy (%)")
    ax[0].legend()
    ax[0].grid(True)

    # Loss
    ax[1].plot(epochs, losses, marker='o')

    ax[1].set_title(f"{plot_title} - Loss")
    ax[1].set_xlabel("Epoch")
    ax[1].set_ylabel("Loss")
    ax[1].grid(True)

    plt.tight_layout()
    plt.show()
```

```python
plot_training_log(
    "./OGB_GNN/results/ogbl-BALBc_no1_spatial_edge_attr_gnn_gcn_20260420234038/log.txt",
    "BALBc1 (spatial edge attributes) | GNN"
)
```
